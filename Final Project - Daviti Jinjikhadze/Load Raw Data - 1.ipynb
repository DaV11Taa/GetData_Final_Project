## 1 - Load Raw Data
ვკითხულობთ Workspace-ში ატვირთულ CSV ფაილს და ვინახავთ getdata.raw.ecommerce_orders-ში როგორც არის - სხვას არ ვცვლით არაფერს

```python
# ვქმნით raw სქემას რათა შემდგომ ჩავწეროთ ინფორმაცია მასში
spark.sql("CREATE SCHEMA IF NOT EXISTS getdata.raw")
```

```python
# წავიკითხოთ CSV ფაილი Workspace-ს path-დან
raw_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Workspace/Users/davitijinjikhadze@gmail.com/Final Project - Daviti Jinjikhadze/ecommerce_orders_dataset.csv") #გაშვებამდე დარწმუნდით რომ ატვირთული CSV ნამდვილად მოცემულ path-ზე მდებარეობს
)

print("Row count:", raw_df.count()) # უნდა იყოს 30000
print("Column count:", len(raw_df.columns)) # უნდა იყოს 41
raw_df.printSchema()
```

```python
raw_df.write.mode("overwrite").saveAsTable("getdata.raw.ecommerce_orders")
print("Saved to getdata.raw.ecommerce_orders")
```
